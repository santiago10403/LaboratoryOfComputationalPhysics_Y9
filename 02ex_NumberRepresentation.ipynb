```python
import math
def roots(a, b, c):
    return ((-b + math.sqrt(b*b - 4 * a * c))/(2 * a),(-b - math.sqrt(b*b - 4 * a * c))/(2 * a))

print(roots(0.001, 1000, 0.001))
```
